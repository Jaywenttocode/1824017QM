```python
import numpy as np
import matplotlib.pyplot as plt 

#defining potential energy function 
def Vpot(x):
    return x**2

a = 0
b = 1
N = 1000

# define x grid and step size 
x = np.linspace(a,b,N)
h = x[1] - x[0]

# creating the kinetic energy matrix 
T =  np.zeros((N-2)**2).reshape(N-2,N-2)
for i in range(N-2):
    for j in range(N-2):
        if i==j:
            T[i,j]= -2
        elif np.abs(i-j)==1:
            T[i,j]=1
        else:
            T[i,j]=0

# creating potential energy matrix 
V = np.zeros((N-2)**2).reshape(N-2,N-2)
for i in range(N-2):
    for j in range(N-2):
        if i==j:
            V[i,j]= Vpot(x[i+1])
        else:
            V[i,j]=0

# create hamiltonian matrix 

H = -T/(2*h**2) + V
print(H)
```
