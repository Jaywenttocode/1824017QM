```python
import numpy as np
import matplotlib.pyplot as plt

hbar = 1
m = 1
l = 0

N = 400                      # Number of grid points
r = np.linspace(0.1, 10, N) # Radial distance r from 0.01 to 10
h = r[1] - r[0]              # Step size

# Effective Potential
V_eff = -1 / r + (hbar**2 * l * (l + 1)) / (2 * m * r**2) 

# H Matrix (H = Kinetic + Potential)
# Kinetic energy uses finite difference for 2nd derivative: (u[i-1] - 2*u[i] + u[i+1]) / h^2
main_diag = -2 * np.ones(N)
off_diag = np.ones(N - 1)

K = (-hbar**2 / (2 * m * h**2)) * (np.diag(main_diag) + np.diag(off_diag, 1) + np.diag(off_diag, -1))
H = K + np.diag(V_eff)


energies, wavefunctions = np.linalg.eigh(H) # Solving for energies (E) and Wavefunctions (u)


plt.figure(figsize=(8, 4))

for n in range(3):
    E = energies[n]
    u = wavefunctions[:, n]
    u = -u
    print(f"State n={n+1}: Energy E = {E:.4f}")
    plt.plot(r, u, label=f"n={n+1}, E={E:.3f}")

# Plot formatting
plt.axhline(0, color='black', linestyle='--')
plt.title("Bound States of Schrödinger Equation for Spherical coordinate")
plt.xlabel("Radius (r)")
plt.ylabel("u(r)")
plt.legend()
plt.grid(True)
plt.show()
```
